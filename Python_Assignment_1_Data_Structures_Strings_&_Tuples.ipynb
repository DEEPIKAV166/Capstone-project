## **String**

### **1.String Concatenation**

```python
string1="Hello "
string2=input("Enter your name: ")
result=string1+string2
print(result)
```

```python
string3=" Welcome to Python programming"
result2=result+string3
print(result2)
```

### **2.String Slicing and Indexing**

```python
print(result2)
```

```python
result2[0]   #Print the first character of the string
```

```python
result2[-1]   #Print the last character of the string
```

```python
print(result2[0:5])    #Print the first 5 characters of the string.
print(result2[-11: ])  #Print the last 11 characters of the string.
print(result2[::-1])   #Print the string in reverse.
print(result2[22:28])  #Use slicing and print the word “Python” from the existing string.
```

### **3.String Methods**

```python
strM = "Python beginner tutorial,"
print(strM)
```

```python
print(strM.upper())   # Convert the sentence to uppercase.
print(strM.lower())   # Convert the sentence to lowercase.
print(strM.capitalize())  # Use Capitalize and return the sentence to the original input form.
print(strM.count("t"))    # Count the total number of occurrences of character ‘t’ in the string.
new_string=strM.replace("Python","Data Analytics") # Replace all occurrences of “Python” with “Data Analytics” in the input string strM = “Python beginner tutorial.”
print(new_string)
```

## **Tuples-Creation, Modification and Access**

```python
first_tuple=(10,20,30) # Create the 1st tuple with values -> (10, 20, 30) and the 2nd tuple with values -> (40, 50, 60):
print(first_tuple)
print(type(first_tuple))

second_tuple=40,50,60
print(second_tuple)
print(type(second_tuple))
```

```python
t_combine=first_tuple+second_tuple # Concatenate the two tuples and store it in “t_combine”
print(t_combine)
print(t_combine*3)    # Repeat the elements of “t_combine” 3 times
print(t_combine[2])   # Access the 3rd element from “t_combine”
print(t_combine[:3])  # Access the first three elements from “t_combine”
print(t_combine[-3:]) # Access the last three elements from “t_combine”
```
