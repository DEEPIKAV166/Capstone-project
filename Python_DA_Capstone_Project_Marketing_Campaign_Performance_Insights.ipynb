```python
# 1. Load the Dataset

import pandas as pd

df = pd.read_csv("/content/Capstone Dataset.csv")
```

```python
# 2. Descriptive Analysis
# Basi Structur:

#  Print the first few rows of the dataset to get an overview of the data.

df.head(10)
```

```python
#  Obtain the number of rows and columns in the dataset.

print("Shape of the dataset:", df.shape)
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

```python
# Get a concise summary of the dataset, including the data types and non-null values.

df.info()
```

```python
# Generate descriptive statistics for numerical columns.

df.describe()
```

```python
# Data Exploratio:
# Print the number of unique Campaign_ID values in the dataset.

df["Campaign_ID"].nunique()
```

```python
# List the unique values of the Location and Customer_Segment columns.

print(df["Location"].nunique())
print(df["Customer_Segment"].nunique())
```

```python
# Count the occurrences of each category in the Campaign_Type and Channel_Used and columns.

print(df["Campaign_Type"].value_counts())
print()
print(df["Channel_Used"].value_counts())
```

```python
# 3. Exploratory Data Analysis (EDA) and Visualization

# Campaig Performanc:

#  Plot a scatter plot to visualize the relationship between Acquisition_Cost and ROI.

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(9,6))
sns.scatterplot(data=df,x="Acquisition_Cost",hue="Campaign_Type",y="ROI",palette="Set2",edgecolor="black",linewidth=0.5,alpha=0.8)
plt.title("Relationship between Acquisition Cost and ROI")
plt.xlabel("Acquisition Cost")
plt.ylabel("ROI")
plt.legend(title="Campaign Type", bbox_to_anchor=(1.02, 1), loc="upper left")
plt.grid(True, linestyle="--", alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
#  Create a bar chart to visualize the average Conversion_Rate for different Channel_Used, categorized by Campaign_Type.

plt.figure(figsize=(10,6))

sns.barplot(data=df, x="Channel_Used", y="Conversion_Rate",hue="Campaign_Type",palette="Set1")
plt.title("Average Conversion Rate by Channel Used and Campaign Type")
plt.xlabel("Channel Used")
plt.ylabel("Average Conversion Rate")
plt.legend(title="Campaign Type", bbox_to_anchor=(1.02,1), loc="upper left")
plt.tight_layout()
plt.show()
```

```python
# Visualize the distribution of Engagement_Score across different Campaign_Type using a box plot.

engagement_grp = df.groupby("Campaign_Type")["Engagement_Score"].apply(list)

plt.figure(figsize=(8,5))
plt.boxplot(
    engagement_grp,
    tick_labels=engagement_grp.index,
    patch_artist=True,
    boxprops=dict(facecolor="lightgreen", color="black"),
    medianprops=dict(color="red"),
    whiskerprops=dict(color="yellow"),
    capprops=dict(color="red"),
    flierprops=dict(marker='o', markerfacecolor='orange')
)

plt.title("Distribution of Engagement Score by Campaign Type")
plt.xlabel("Campaign Type")
plt.ylabel("Engagement Score")
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
# Analyze the average ROI by Company using a bar chart to compare the profitability of campaigns conducted by different companies.

roi_grp = df.groupby("Company")["ROI"].mean().reset_index()

plt.figure(figsize=(10,5))
sns.barplot(data=roi_grp, x="Company", y="ROI",color="skyblue",edgecolor="black")

plt.title("Average ROI by Company")
plt.xlabel("Company")
plt.ylabel("Average ROI")
#plt.xticks(rotation=45)
plt.grid(axis="y", linestyle="--", alpha=0.7)

plt.tight_layout()
plt.show()
```

```python
#  Examine the correlation between Engagement_Score and Conversion_Rate using a heatmap

corr = df[["Engagement_Score", "Conversion_Rate"]].corr()

plt.figure(figsize=(5,4))
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")

plt.title("Correlation between Engagement Score and Conversion Rate")

plt.tight_layout()
plt.show()
```

```python
# Customer Segmentatio:

# Create a count plot to visualize the distribution of Target_Audience.

plt.figure(figsize=(8,6))
sns.countplot(x = "Target_Audience", data = df,color="orange",edgecolor="black")
plt.title("Distribution of Target Audience")
plt.xlabel("Target Audience")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

```python
# Identify which Customer_Segment has the highest Conversion_Rate for each Language using a bar chart.

plt.figure(figsize=(8,6))

sns.barplot(x="Customer_Segment",
           y="Conversion_Rate",
           hue="Language",
           errorbar=None,
           palette="dark:pink",
           edgecolor="gray",
           data=df
            )
plt.title("Conversion Rate by Customer Segment and Language")
plt.xlabel("Customer Segment")
plt.ylabel("Conversion Rate")
plt.legend(bbox_to_anchor=(1,1))
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
#  Visualize the distribution of Acquisition_Cost across each Customer_Segment,categorized by Channel_Used, using a box plot.

plt.figure(figsize=(10,6))
sns.boxplot(
    x="Acquisition_Cost",
    y="Customer_Segment",
    hue="Channel_Used",
    data=df,
    showfliers=True,
    palette="bright",
    orient="h"
)
plt.title("Distribution of Acquisition Cost by Customer Segment and Channel")
plt.xlabel("Acquisition Cost")
plt.ylabel("Customer Segment")
plt.legend(bbox_to_anchor=(1,1))
plt.tight_layout()
plt.show()
```

```python
#  Analyze average Conversion_Rate by Language using a bar chart to compare the effectiveness of campaigns conducted in different languages.

plt.figure(figsize=(10,6))

sns.barplot(x="Conversion_Rate",
           y="Language",
           hue="Language",
           palette="coolwarm",
           edgecolor="black",
           estimator="mean",
           errorbar=None,
           data=df,
           legend=False
            )
plt.title("Average Conversion Rate by Language")
plt.xlabel("Average Conversion Rate")
plt.ylabel("Language")
plt.tight_layout()
plt.show()
```

```python
# Channe Effectivenes:
# Compare the Engagement_Score for different Channels_Used, segmented by Campaign_Type, using a bar chart.

plt.figure(figsize=(10,6))

sns.barplot(x="Channel_Used",
           y="Engagement_Score",
           hue="Campaign_Type",
           palette="viridis",
           edgecolor="black",
           estimator="mean",
           errorbar=None,
           data=df
            )
plt.legend(title="Campaign Type",
    bbox_to_anchor=(1.02, 1),
    loc="upper left")
plt.title("Average Engagement Score by Channel and Campaign Type")
plt.xlabel("Channel Used")
plt.ylabel("Average Engagement Score")
plt.tight_layout()
plt.show()
```

```python
# Show the distribution of total ROI across different Channels_Used using a pie chart

channel_roi = df.groupby("Channel_Used")["ROI"].sum()

plt.figure(figsize=(8, 6))
plt.pie(
    channel_roi,
    colors = ['skyblue', 'lightgreen', 'pink', 'violet', 'orange','yellow'],
    labels=channel_roi.index,
    autopct="%1.1f%%",
    startangle=90,
    explode=[0, 0, 0.1, 0.1,0, 0],
    shadow=True
)

plt.title("Distribution of Total ROI by Channel")
plt.show()
```

```python
#  Plot a scatter plot to show the relationship between Clicks and Impressions for each Channel_Used.

plt.figure(figsize=(10,6))
sns.scatterplot(
    data=df,
    x='Clicks',
    y='Impressions',
    hue='Channel_Used',
    palette='Set1',
    marker='P')

plt.title("Relationship Between Clicks and Impressions by Channel")
plt.xlabel("Clicks")
plt.ylabel("Impressions")
plt.legend(bbox_to_anchor=(1, 1), title="Channel Used")

plt.tight_layout()
plt.show()
```

```python
# Time-Base Analysi:

# Plot the distribution of Duration using a histogram.

plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="Duration",
    hue="Channel_Used",
    bins=10,
    multiple="dodge",
    palette="Set3",
    stat="count",
    kde=True
)

plt.title("Distribution of Duration")
plt.xlabel("Duration")
plt.ylabel("Count")

plt.tight_layout()
plt.show()
```

```python
# Analyze how the overall Conversion_Rate has changed over Date for each Company using a line chart.

df["Date"] = pd.to_datetime(df["Date"],dayfirst=True)

conversion_trend = (df.groupby(["Date", "Company"])["Conversion_Rate"].mean().reset_index())

plt.figure(figsize=(12, 6))

sns.lineplot(data=conversion_trend,x="Date",y="Conversion_Rate",hue="Company")

plt.title("Conversion Rate Trend Over Time by Company")
plt.xlabel("Date")
plt.ylabel("Average Conversion Rate")
plt.legend(title="Company", bbox_to_anchor=(1, 1), loc="upper left")
plt.tight_layout()
plt.show()
```

```python
# Examine the trend of Engagement_Score over Date with a line chart.

df["Date"] = pd.to_datetime(df["Date"], format="%d-%m-%Y")

engagement_trend = (df.groupby("Date")["Engagement_Score"].mean().reset_index())

plt.figure(figsize=(10,6))

sns.lineplot(data=engagement_trend,x="Date",y="Engagement_Score")
plt.title("Trend of Engagement Score Over Time")
plt.xlabel("Date")
plt.ylabel("Average Engagement Score")
plt.tight_layout()
plt.show()
```

```python
# Geographi Analysi:

# Determine which location has the highest Acquisition_Cost using a bar chart.

acquisition_by_location = (df.groupby("Location")["Acquisition_Cost"].mean().reset_index().sort_values("Acquisition_Cost", ascending=False))

plt.figure(figsize=(12, 6))

sns.barplot(data=acquisition_by_location,x="Location",y="Acquisition_Cost",palette="Set2",hue="Location",legend=False)

plt.title("Average Acquisition Cost by Location")
plt.xlabel("Location")
plt.ylabel("Average Acquisition Cost")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
#  Visualize the Conversion_Rate by different Location, categorized by Target_Audience, using a bar chart.

plt.figure(figsize=(12, 6))

sns.barplot(data=df,x="Location",y="Conversion_Rate",hue="Target_Audience",palette="Set1")

plt.title("Conversion Rate by Location and Target Audience")
plt.xlabel("Location")
plt.ylabel("Average Conversion Rate")
plt.legend(title="Target Audience", bbox_to_anchor=(1.02, 1), loc="upper left")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
#  Illustrate the proportion of ROI by Location using a pie chart.

roi_by_location = (df.groupby("Location")["ROI"].sum().reset_index())

plt.figure(figsize=(8, 8))
plt.pie(roi_by_location["ROI"],labels=roi_by_location["Location"],autopct="%1.1f%%",startangle=90)
plt.title("Proportion of ROI by Location")
plt.show()
```

# Project Summary – Marketing Campaign Performance Insights

## Key Insights

The dataset contains **22,029 marketing campaigns and 16 columns**, with no missing values in the loaded data. It includes information about companies, campaign types, target audiences, channels, conversion rates, acquisition costs, ROI, engagement scores, customer segments, locations, languages, clicks, impressions, duration, and dates.

There are **10 locations and 5 customer segments**. Among the campaign types, **Display** has the highest number of campaigns (4,450), followed by **Search** (4,441), while **Influencer** has the lowest (4,338). Among the channels, **Facebook** has the highest number of campaigns (3,742), while **Email** has the lowest (3,624).

The overall **average Conversion Rate is approximately 4.76%**, while the average **Engagement Score is approximately 6.58**. The average **Acquisition Cost is approximately 5,522.74**, and the average **ROI is approximately 182.86**. The wide variation in ROI indicates that campaign profitability differs considerably across campaigns.

## Analysis Performed

The project began with **descriptive analysis** to understand the structure and characteristics of the dataset. The number of rows and columns, data types, missing values, unique Campaign IDs, locations, customer segments, campaign types, and channels were examined. Descriptive statistics were also generated for numerical variables such as Conversion Rate, Acquisition Cost, ROI, Clicks, Impressions, and Engagement Score.

**Exploratory Data Analysis (EDA)** was performed using different visualization techniques. Scatter plots were used to examine the relationships between **Acquisition Cost and ROI** and between **Clicks and Impressions**. Bar charts were used to compare **Conversion Rate, ROI, and Engagement Score** across different channels, companies, languages, campaign types, and customer segments. Box plots were used to analyze the distribution of **Engagement Score and Acquisition Cost** across different campaign categories. A heatmap was used to examine the correlation between **Engagement Score and Conversion Rate**.

The **customer segmentation analysis** examined how different target audiences, customer segments, and languages respond to marketing campaigns. Conversion Rate was compared across customer segments and languages, while Acquisition Cost was analyzed across customer segments and channels.

The **channel effectiveness analysis** compared Engagement Score and ROI across different marketing channels. The relationship between Clicks and Impressions was also analyzed to understand user interaction and channel performance.

The **time-based analysis** examined campaign performance over time. Line charts were used to analyze changes in **Conversion Rate for each Company** and the trend of **Engagement Score over Date**. A histogram was used to examine the distribution of campaign Duration.

The **geographical analysis** examined the influence of Location on campaign performance. Acquisition Cost was compared across locations, Conversion Rate was analyzed by Location and Target Audience, and the proportion of ROI contributed by different locations was visualized using a pie chart.

## Overall Conclusion

Overall, this project provides a comprehensive view of **marketing campaign performance across companies, campaign types, channels, customer segments, languages, time periods, and locations**. The analysis helps identify differences in campaign effectiveness, customer engagement, acquisition costs, and profitability. These insights can support **data-driven marketing decisions, better channel selection, improved customer targeting, cost optimization, and higher ROI**.
